```python
import numpy as np
import matplotlib.pyplot as plt

city =np.array(["Tehran" , "Shiraz" , "Esfehan" ,"Kashan" , "Tabriz" , "Yazd" , "Mashhad" ,"Kerman" , "Rasht" , "Ramsar"])

population = np.array([8693706,1565572,1961260,304000,1558693,529673,3001184,537718,679995,35000])

Longitude = np.array([51.42151, 52.53113, 51.67462, 51.41001, 46.29190,
                      54.36750, 59.60567, 57.07879, 49.58862, 50.65858])

latitude = np.array([35.69439, 29.61031, 32.65246, 33.98504, 38.08000,
                    31.89722, 36.29807, 30.28321, 37.27611, 36.90336])
```

```python
plt.bar(city , population )
plt.xlabel("city")
plt.ylabel("population")
plt.xticks(rotation = 45)
plt.title("Bar")
plt.show()
```

```python
plt.plot(city , population)
plt.xlabel("City")
plt.ylabel("Population")
plt.xticks(rotation = 45)
plt.title("Bar")
plt.show()
```

```python
for i in range(len(city)):
    plt.text(Longitude[i] , latitude[i] , city[i])
plt.scatter(Longitude , latitude , population/10000)
plt.xlabel("latitude")
plt.ylabel("Longitude")
plt.title("Scatter")
plt.show()
```

```python
plt.pie(population , labels = city , autopct = "%1.1f%%")
plt.title("Pie")
plt.show()
```
